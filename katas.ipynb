1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def diccionario_letras(texto):
    diccionario = {}
    for letra in texto:
        if letra.isalpha():  # Solo letras (sin espacios ni símbolos)
            letra = letra.lower()  # Pasar todo a minúsculas
            if letra in diccionario:
                diccionario[letra] += 1 #si ya existe, suma 1
            else:
                diccionario[letra] = 1 #si no existe, lo crea con valor 1
    return diccionario

texto = "Esto es un texto de prueba para contar letras." 
print(diccionario_letras(texto))           
```

6. Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(n):
    # Caso base
    if n == 0 or n == 1:
        return 1
    
    # Caso recursivo
    return n * factorial(n - 1)

print(factorial(5))  

```

14.Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

```python
# Función que devuelve las palabras que empiezan por una letra concreta
def retornar_palabras(lista_palabra, letra):

    # Función que comprueba si una palabra empieza por la letra indicada
    def empieza_por(palabra):
        # Convertimos ambas a minúsculas para evitar problemas con mayúsculas
        return palabra.lower().startswith(letra.lower())

    # filter recorre la lista y solo deja pasar las palabras
    # para las que empieza_por devuelve True
    resultado = list(filter(empieza_por, lista_palabra))

    # Devolvemos la lista filtrada
    return resultado


palabras = ["Casa", "Coche", "Perro", "Camión", "Gato"]

print(retornar_palabras(palabras, "C"))

```

28. Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
def primer_duplicado(lista):
    vistos = set()  # Conjunto para ir guardando los elementos que ya hemos visto

    for elemento in lista:
        if elemento in vistos:
            return elemento  # Si ya lo vimos, es el primer duplicado
        vistos.add(elemento)  # Si no lo vimos, lo añadimos al conjunto

    return "No hay elementos repetidos"  # Si no hay duplicados, devolvemos None


lista_ejemplo = [2, 5, 1, 2, 3, 5, 1]
lista_ejemplo2 = [2, 5, 1, 3]
print(f"El primer duplicado es: {primer_duplicado(lista_ejemplo)}")
print(f"El primer duplicado es: {primer_duplicado(lista_ejemplo2)}")
```

29. Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.

```python
def enmascarar_variable(valor):
    # Convertimos la variable a string
    texto = str(valor)
    
    # Calculamos cuántos caracteres hay que reemplazar con #
    num_enmascarar = len(texto) - 4
    
    if num_enmascarar > 0:
        # Reemplazamos todos los caracteres excepto los últimos cuatro
        return '#' * num_enmascarar + texto[-4:]
    else:
        # Si la longitud es menor o igual a 4, devolvemos la cadena tal cual
        return texto

# Ejemplos de uso
print(enmascarar_variable(1234567890))   # ######7890
print(enmascarar_variable("contraseña")) # #####seña
print(enmascarar_variable(123))          # 123


```

30. Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
def son_anagramas(palabra1, palabra2):
    # Pasamos ambas palabras a minúsculas y eliminamos espacios
    p1 = palabra1.replace(" ", "").lower()
    p2 = palabra2.replace(" ", "").lower()

    # Comparamos las letras ordenadas / sorted() convierte la palabra en una lista de letras ordenadas alfabéticamente
    return sorted(p1) == sorted(p2)

print(son_anagramas("Roma", "Amor"))        
print(son_anagramas("Hola", "Adiós"))      
```

34. Crea la clase Arbol.

Define un árbol genérico con un tronco y ramas como atributos.

Métodos disponibles:
- crecer_tronco
- nueva_rama
- crecer_ramas
- quitar_rama
- info_arbol

Código a seguir:
- Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
- Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.
- Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
- Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
- Implementar el método quitar_rama para eliminar una rama en una posición específica.
- Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.

Caso de uso:
-  Crear un árbol.
-  Hacer crecer el tronco una unidad.
-  Añadir una nueva rama.
-  Hacer crecer todas las ramas una unidad.
-  Añadir dos nuevas ramas.
-  Retirar la rama situada en la posición 2.
-  Obtener información sobre el árbol.


```python
class Arbol:
    # Inicializar el árbol
    def __init__(self): 
        self.tronco = 1 # Longitud 1
        self.ramas = [] # Lista vacía de ramas

    # Crecer el tronco: aumentamos su longitud en una unidad
    def crecer_tronco(self):
        self.tronco += 1    

    # Añadir nueva rama: agregamos una nueva rama de longitud 1 a la lista de ramas
    def nueva_rama(self):
        self.ramas.append(1)    

    # Hacer crecer todas las ramas: aumentamos en 1 unidad la longitud de todas las ramas existentes
    def crecer_ramas(self):    
        for i in range(len(self.ramas)):
            self.ramas[i] += 1

    # Quitar rama: eliminar una rama en una posición específica
    def quitar_rama(self, posicion):
        if 0 <= posicion < len(self.ramas): # Comprobamos la posición: no puede ser negativa ni mayor al número de ramas existentes
            self.ramas.pop(posicion)

    # Información del árbol: devuelve longitud del tronco, número de ramas y sus longitudes
    def info_arbol(self):    
        return {
            "longitud_tronco": self.tronco,
            "numero_ramas": len(self.ramas),
            "ramas": self.ramas
        }    

# Caso de uso

# a. Crear un árbol
arbol = Arbol()

# b. Crecer el tronco
arbol.crecer_tronco()

# c. Añadir una nueva rama
arbol.nueva_rama()

# d. Crecer todas las ramas
arbol.crecer_ramas()

# e. Añadir dos nuevas ramas
arbol.nueva_rama()
arbol.nueva_rama()

# f. Quitar la rama en la posición 2
arbol.quitar_rama(2)

# g. Mostrar información
print(arbol.info_arbol())
```

35. Crea la clase UsuarioBanco
- Representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente.
- Métodos: retirar_dinero, transferir_dinero, agregar_dinero.
- Código a seguir:
-  Inicializar un usuario con nombre, saldo y un indicador (True o False) de cuenta corriente.
-  Implementar retirar_dinero para sustraer dinero del saldo, lanzando un error si no es posible.
- Implementar transferir_dinero para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
- Implementar agregar_dinero para aumentar el saldo del usuario.
Caso de uso:
-  Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
-  Agregar 20 unidades al saldo de Bob.
 -  Transferir 80 unidades de Bob a Alicia.
 - Retirar 50 unidades del saldo de Alicia.

```python
class usuario_banco:
    def __init__(self, nombre, saldo, cuenta_corriente):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    # Agregar dinero
    def agregar_dinero(self, cantidad):
        if cantidad <= 0:
            raise ValueError("La cantidad debe de ser positiva")   
        self.saldo += cantidad 

    # Retirar dinero
    def retirar_dinero(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("Saldo insuficiente")    
        self.saldo -= cantidad

    # Transferir dinero
    def transferir_dinero(self, otro_usuario, cantidad):
        if cantidad > self.saldo:
            raise ValueError("No hay saldo suficiente para transferir")
        self.saldo -= cantidad # Quien envia el dinero
        otro_usuario.saldo += cantidad # Quien lo recibe

# Caso de uso

# a
alicia = usuario_banco("Alicia", 100, True)    
bob = usuario_banco("Bob", 50, True)

# b
bob.agregar_dinero(20)

# c
bob.transferir_dinero(alicia, 80)

# d
alicia.retirar_dinero(50)
```

Procesa un texto según la opción especificada: contar_palabras, reemplazar_palabras o eliminar_palabra.

Código a seguir:

- Crear una función contar_palabras que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
- Crear una función reemplazar_palabras para sustituir una palabra_original por una palabra_nueva en el texto y devolver el texto modificado.
- Crear una función eliminar_palabra que elimine una palabra del texto y devuelva el texto sin ella.
- Crear la función procesar_texto que reciba un texto, una opción ("contar", "reemplazar", "eliminar") y un número variable de argumentos según la  opción elegida.

Caso de uso:
- Verificar el funcionamiento completo de procesar_texto.

```python
# 1. Funcion contar palabras

def contar_palabras(texto):
    palabras = texto.lower().split()
    contador = {} # Diccionario vacio

    for palabra in palabras:
        if palabra in contador:
            contador[palabra] += 1
        else:
            contador[palabra] = 1    
    return contador

# 2. Funcion reemplazar palabras

def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    return texto.replace(palabra_original, palabra_nueva)

# 3. Eliminar palabras

def eliminar_palabra(texto, palabra_a_eliminar):
    palabras = texto.split()
    palabras_filtradas = [] # Lista nueva sin la palabra prohibida

    for palabra in palabras:
        if palabra != palabra_a_eliminar:
            palabras_filtradas.append(palabra) # Guardamos las palabras que no sean las que queremos borrar

    return " ".join(palabras_filtradas) # Convierte la lista en texto otra vez

# 4. Procesar texto  

def procesar_texto(texto, opcion, *args):
    if opcion == "contar":
        return contar_palabras(texto)

    elif opcion == "reemplazar":
        return reemplazar_palabras(texto, args[0], args[1])

    elif opcion == "eliminar":
        return eliminar_palabra(texto, args[0])

    else:
        raise ValueError("Opción no válida")
    

# Caso de uso

texto = "El paseo de hoy ha sido largo"

# Contar palabras
print(procesar_texto(texto, "contar"))

# Reemplazar palabras
print(procesar_texto(texto, "reemplazar", "largo", "corto"))

# Eliminar palabra
print(procesar_texto(texto, "eliminar", "hoy"))


```

38. Escribe un programa que determine qué calificación en texto tiene un alumno según su calificación numérica.

Reglas:

        0 - 69: insuficiente

        70 - 79: bien

        80 - 89: muy bien
        
        90 - 100: excelente

```python
def calificacion(nota):
    if not isinstance(nota, int):
        raise TypeError("La nota tiene que ser un número")
    
    if nota < 0 or nota > 100:
        raise ValueError("La nota tiene que estar entre 0 y 100")
    
    elif 0 <= nota <= 69:
        print("Insuficiente")  
    elif 70 <= nota <= 79:
        print("Bien")
    elif 80 <= nota <= 89:
        print("Muy bien")    
    else:
        print("Excelente") 

calificacion(20) 
calificacion(100)
calificacion(89)
calificacion(125)   
           
```

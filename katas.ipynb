1. Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def diccionario_letras(texto):
    diccionario = {}
    for letra in texto:
        if letra.isalpha():  # Solo letras (sin espacios ni símbolos)
            letra = letra.lower()  # Pasar todo a minúsculas
            if letra in diccionario:
                diccionario[letra] += 1 #si ya existe, suma 1
            else:
                diccionario[letra] = 1 #si no existe, lo crea con valor 1
    return diccionario

texto = "Esto es un texto de prueba para contar letras." 
print(diccionario_letras(texto))           
```

3. Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
def filtrar_palabras(lista, palabra_objetivo):
  lista_objetivo = []

  for palabra in lista:
    if palabra_objetivo in palabra:
      lista_objetivo.append(palabra)
  return lista_objetivo    

lista_palabras = ["aguacate", "banana", "cereza", "dátil", "arándano"] 
palabra_objetivo = "a" 

print(filtrar_palabras(lista_palabras, palabra_objetivo))
```

5. Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
lista_tres = [5, 9, 2, 4, 1]

def media_numeros(lista_numeros, nota_aprobado=5):
    media = sum(lista_numeros) / len(lista_numeros)
    if media >= nota_aprobado:
        estado = "Aprobado"
    else:
        estado = "Suspenso" 

    return media, estado    

print(media_numeros(lista_tres))
```

6. Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(n):
    # Caso base
    if n == 0 or n == 1:
        return 1
    
    # Caso recursivo
    return n * factorial(n - 1)

print(factorial(5))  

```

9. Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().

```python
mascotas_prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]

#Función que decide si una mascota es válida
def mascota_permitida(mascota):
    return mascota not in mascotas_prohibidas

def filtrar_mascotas(nombres_mascotas):
    #filter = filtra una lista usando una función que devuelve True o False
    return list(filter(mascota_permitida, nombres_mascotas))


nombres_mascotas = ["Perro", "Gato", "Mapache", "Loro", "Tigre"]
print(filtrar_mascotas(nombres_mascotas))

```

10. Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.

```python
lista_numeros = [2, 1, 9, 43, 67]

def calcular_promedio(lista_numeros):
    try: 
        if len(lista_numeros) == 0:
            raise ValueError("La lista está vacía, no se puede calcular el promedio.")

        # Calculamos el promedio    
        promedio = sum(lista_numeros) / len(lista_numeros)
        return promedio
    
    # Capturamos la excepción personalizada
    except ValueError as e:
        print(f"Error: {e}")  

# Probamos con una lista no vacía
print(calcular_promedio(lista_numeros))  

# Probamos con una lista vacía
print(calcular_promedio([]))         
```

12. Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().

```python
frase = "LLeva demasiados dias lloviendo"

def logitud_palabra(frase):
    palabras = frase.split() # Separa la frase en palabras
    longitudes = list(map(len, palabras))  # Aplica len a cada palabra
    return longitudes

print(logitud_palabra(frase))

```

13. Genera una función que, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas. Las letras no pueden estar repetidas. Usa la función map().

```python
def letras_mayus_minus(caracteres):
    letras_unicas = set(caracteres) #Elimina letras repetidas

    def convertir(letra):
        return (letra.upper(), letra.lower())

    resultado = list(map(convertir, letras_unicas))
    return resultado

print(letras_mayus_minus("tgButyv"))
```

14.Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

```python
# Función que devuelve las palabras que empiezan por una letra concreta
def retornar_palabras(lista_palabra, letra):

    # Función que comprueba si una palabra empieza por la letra indicada
    def empieza_por(palabra):
        # Convertimos ambas a minúsculas para evitar problemas con mayúsculas
        return palabra.lower().startswith(letra.lower())

    # filter recorre la lista y solo deja pasar las palabras
    # para las que empieza_por devuelve True
    resultado = list(filter(empieza_por, lista_palabra))

    # Devolvemos la lista filtrada
    return resultado


palabras = ["Casa", "Coche", "Perro", "Camión", "Gato"]

print(retornar_palabras(palabras, "C"))

```

16. Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las palabras que sean más largas que n. Usa la función filter().

```python
def retornar_palabras_largas(cadena_texto, n):

    # Convertimos la cadena en una lista de palabras

    palabras = cadena_texto.split()

    # Función que comprueba si una palabra es más larga que n
    def comprueba_palabras(palabra):
        return len(palabra) > n
    
    # Filtramos las palabras que cumplen la condición
    resultado = list(filter(comprueba_palabras, palabras))

    return resultado

texto = "Mi perro se llama Bali"
print(retornar_palabras_largas(texto, 3))
```

17. Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().

```python
# Importamos reduce desde functools
from functools import reduce

# Función que une los dígitos paso a paso
def unir_digitos(acc, digito):
    # Multiplicamos el número acumulado por 10
    # y sumamos el nuevo dígito
    return acc * 10 + digito

# Función principal que recibe la lista de dígitos
def funcion_digitos(lista_digitos):
    # reduce aplica la función unir_digitos
    # a todos los elementos de la lista
    return reduce(unir_digitos, lista_digitos)

reduce(unir_digitos, [5, 7, 2])
```
